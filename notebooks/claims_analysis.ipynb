```python
import pandas as pd
import seaborn as sn
import matplotlib.pyplot as plt
```

```python
## PART 1. DATA LOADING AND EXPLORATION
```

```python
#loading csv files to dataframe variable
df_header = pd.read_csv('data/STONYBRK_20240531_HEADER.csv', encoding='utf-8', on_bad_lines='warn')
df_code = pd.read_csv('data/STONYBRK_20240531_CODE.csv', encoding='utf-8', on_bad_lines='warn')
df_line = pd.read_csv('data/STONYBRK_20240531_LINE.csv', encoding='utf-8', on_bad_lines='warn')
```

```python
#displaying number of rows and columns using shape
print("df_header shape:", df_header.shape)
print("df_code shape:", df_code.shape)
print("df_line shape:", df_line.shape)
```

```python
#displaing first 5 rows of each df
print("df_header - First 5 rows:")
display(df_header.head())

print("df_code - First 5 rows:")
display(df_code.head())

print("df_line - First 5 rows:")
display(df_line.head())
```

```python
#dispalaying column names and  data types
print("df_header - Column names and data types:")
print(df_header.dtypes)

print("\ndf_code - Column names and data types:")
print(df_code.dtypes)

print("\ndf_line - Column names and data types:")
print(df_line.dtypes)
```

```python
#displaying missing value and its count
print("df_header - Missing values:")
print(df_header.isnull().sum())

print("\ndf_code - Missing values:")
print(df_code.isnull().sum())

print("\ndf_line - Missing values:")
print(df_line.isnull().sum())
```

```python
#displaying descriptive statistics for numeric columns
print("df_header - Descriptive statistics:")
print(df_header.describe())

print("\ndf_code - Descriptive statistics:")
print(df_code.describe())

print("\ndf_line - Descriptive statistics:")
print(df_line.describe())
```

```python
unique_claims_head = df_header['ProspectiveClaimId'].nunique()
unique_claims_code = df_code['ProspectiveClaimId'].nunique()
unique_claims_line = df_line['ProspectiveClaimId'].nunique()

print("Unique claims in df_header:", unique_claims_head)
print("Unique claims in df_code:", unique_claims_code)
print("Unique claims in df_line:", unique_claims_line)
```

```python
# Displaying the date range of claims
# Convert columns to datetime format
df_header['ServiceFromDate'] = pd.to_datetime(df_header['ServiceFromDate'])
df_header['ServiceToDate'] = pd.to_datetime(df_header['ServiceToDate'])

print(f"Service Date Range from: {df_header['ServiceFromDate'].min().date()} to {df_header['ServiceToDate'].max().date()}")
```

```python
# Group by claim and count service lines
service_line_counts = df_line.groupby('ProspectiveClaimId').size()

# Calculate the average
average_service_lines = service_line_counts.mean()

print(f"Average number of service lines per claim: {average_service_lines:.2f}")
```

```python
# Group by Prospectiveclaimid and count the number of CodeValue entries per claim
code_counts = df_code.groupby('ProspectiveClaimId')['CodeValue'].count()

# Calculate the average number of diagnosis codes per claim
average_codes_per_claim = code_counts.mean()

print(f"Average diagnosis codes per claim: {average_codes_per_claim:.2f}")
```

```python
## PART 2. RELATIONAL DATA ANALYSIS
```

```python
##Question 1. Provider Analysis
#Who are the top 5 billing providers by number of claims?
top5_billingproviders = (
    df_header.groupby('BillingProvFirstName')['ProspectiveClaimId']
    .nunique()  # Count unique claims per provider
    .sort_values(ascending=False)  # Sort descending
    .head(5) ) # Get top 5


print(top5_billingproviders)
```

```python
# Display: Provider name, NPI, and claim count
top_providers = (
    df_header.groupby(['BillingProvFirstName', 'BillingProviderNPI'])['ProspectiveClaimId']
    .nunique()
    .reset_index(name='ClaimCount')
    .sort_values(by='ClaimCount', ascending=False)  )

print(top_providers)
```

```python
# Create a simple bar chart showing the top 5 providers
# Step 2: Plot the results
plt.figure(figsize=(8, 5))
top5_billingproviders.plot(kind='barh', color='steelblue')
plt.xlabel('Number of Unique Claims')
plt.ylabel('Billing Provider')
plt.title('Top 5 Billing Providers by Number of Claims')
plt.gca().invert_yaxis()  # Highest bar at the top
plt.tight_layout()
plt.show()
```

```python
#Question 2. Pyer Mix Analysis
#What are the top 5 primary payers by claim volume?
top5_payers = (
    df_header.groupby('PrimaryPayerName')['ProspectiveClaimId']
    .nunique()
    .sort_values(ascending=False)
    .head(5)  )

print(top5_payers)
```

```python
#Calculate the percentage of total claims for each payer
payers = (
    df_header.groupby('PrimaryPayerName')['ProspectiveClaimId']
    .nunique()
    .sort_values(ascending=False) )

total_claims = df_header['ProspectiveClaimId'].nunique()

# Step 3: Convert to DataFrame and calculate percentage
payers_df = payers.reset_index(name='ClaimCount')
payers_df['ClaimPercent'] = (payers_df['ClaimCount'] / total_claims * 100).round(2)

print(payers_df)
```

```python
# Sort by claim count descending (already sorted, but safe to ensure)
payers_df_sorted = payers_df.sort_values(by='ClaimCount', ascending=False)

# Plot
plt.figure(figsize=(10, 6))
plt.barh(payers_df_sorted['PrimaryPayerName'], payers_df_sorted['ClaimCount'], color='teal')
plt.xlabel('Number of Unique Claims')
plt.ylabel('Primary Payer Name')
plt.title('Claim Volume by Primary Payer')
plt.gca().invert_yaxis()  # Highest bar at the top
plt.tight_layout()
plt.show()
```

```python
#Question 3. Common Diagnosis
#What are the 10 most frequently appearing diagnosis codes (CodeValue)?
top_10_codes = (
    df_code['CodeValue']
    .value_counts()
    .head(10) )

print(top_10_codes)
```

```python
#Display: ICD-10 code and frequency count
top_10_codes = (
    df_code['CodeValue']
    .value_counts()
    .reset_index()
    .head(10)  )

# Rename columns directly (no "index" left over)
top_10_codes.columns = ['ICD-10 Code', 'Frequency']

# Dictionary mapping ICD-10 codes to descriptions
icd10_descriptions = {
    "J96.01": "Acute respiratory failure with hypoxia",
    "I10": "Essential (primary) hypertension",
    "E78.5": "Hyperlipidemia, unspecified",
    "G93.5": "Compression of brain",
    "D64.9": "Anemia, unspecified",
    "I25.10": "Atherosclerotic heart disease of native coronary artery without angina pectoris",
    "I61.9": "Nontraumatic intracerebral hemorrhage, unspecified",
    "I48.91": "Unspecified atrial fibrillation",
    "I60.8": "Other nontraumatic subarachnoid hemorrhage",
    "I50.9": "Heart failure, unspecified" }

# Add descriptions to the DataFrame
top_10_codes['Description'] = top_10_codes['ICD-10 Code'].map(icd10_descriptions)

print(top_10_codes)
```

```python
#What are the 10 most frequently billed procedure codes (HCPCS)?
top_10_hcpcs = (
    df_line['HCPCS']              # <-- using df_line and HCPS column
    .value_counts()
    .reset_index()
    .head(10)  )

# Rename columns for clarity
top_10_hcpcs.columns = ['HCPCS Code', 'Frequency']

# Display results
print(top_10_hcpcs)
```

```python
#Display: HCPCS code, description (if available in data), and frequency
top_10_hcpcs = (
    df_line
    .groupby(['HCPCS', 'ClientProcedureName'])   # group by code + description
    .size()                                     # count occurrences
    .reset_index(name='Frequency')              # turn into DataFrame
    .sort_values(by='Frequency', ascending=False)  # sort by frequency
    .head(10)  )                                 # top 10


# Reorder columns: HCPCS -> Frequency -> Description
top_10_hcpcs = top_10_hcpcs[['HCPCS', 'Frequency', 'ClientProcedureName']]

# Display results
print(top_10_hcpcs)
```

```python
#Create a bar chart showing the top 10 procedures #base on df above
plt.figure(figsize=(10,6))
plt.bar(top_10_hcpcs['HCPCS'], top_10_hcpcs['Frequency'], color='skyblue')

plt.title("Top 10 HCPCS Procedure Codes")
plt.xlabel("HCPCS Code")
plt.ylabel("Frequency")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Question 5. Service Location Analysis
#How many claims were submitted for each PlaceOfService?
service_counts = (
    df_header
    .groupby('PlaceOfService')['ProspectiveClaimId']   # group by both columns
    .count()                                                            # count claims
    .reset_index(name='ClaimCount')                                     # turn into DataFrame
    .sort_values(by='ClaimCount', ascending=False) )                      # optional: sort by count

# Display results
print(service_counts)
```

```python
#What percentage of claims are for "INPATIENT" vs "DOCTOR'S OFFICE"?

# Count claims by Facility_Location
facility_counts = (
    df_header
    .groupby('Facility_Location')['ProspectiveClaimId']
    .count()
    .reset_index(name='ClaimCount'))

# Calculate percentages
facility_counts['Percentage'] = (
    facility_counts['ClaimCount'] / facility_counts['ClaimCount'].sum() * 100
).round(2)

# Filter only INPATIENT vs DOCTOR'S OFFICE
facility_subset = facility_counts[facility_counts['Facility_Location'].isin(['INPATIENT', "DOCTOR'S OFFICE"])]

print(facility_subset)
```

```python
## Part 3: ADVANCE ANALYSIS WITH JOINS
```

```python
#Question 6: Claims with High Service Line Counts
#merging 2 dataframe df_header and df_line
headerline_df = pd.merge(df_header, df_line, on='ProspectiveClaimId', how='inner')
print(headerline_df)

```

```python
#Calculate the total number of service lines per claim
service_lines_per_claim = (
    headerline_df
    .groupby('ProspectiveClaimId')   # group by claim ID
    .size()                          # count rows (service lines)
    .reset_index(name='ServiceLineCount')  # turn into DataFrame
    .sort_values(by='ServiceLineCount', ascending=False)  # optional: sort
)

# Display results
print(service_lines_per_claim)
```

```python
#Identify claims with 5 or more service lines
claims_with_5_or_more = service_lines_per_claim[
    service_lines_per_claim['ServiceLineCount'] >= 5
]

# Display results
print(claims_with_5_or_more)
```

```python
# Display: ClaimId, Provider name, number of lines, and total charges
claim_summary = (
    headerline_df
    .groupby(['ProspectiveClaimId', 'BillingProvFirstName'])
    .agg(
        LinePos=('ProspectiveClaimId', 'count'),   # number of service lines
        TotalCharges=('Charges', 'sum')            # total charges
    )
    .reset_index()
    .sort_values(by='LinePos', ascending=False)    # optional: sort by number of lines
)

# Display results
print(claim_summary)
```

```python
#Question 7: Diagnosis-Procedure Combinations
```

```python
#Create a merged dataset linking claims to both procedures and diagnoses

# First merge header with code and line
merge_df = (
    df_header
    .merge(df_code, on='ProspectiveClaimId', how='left')
    .merge(df_line, on='ProspectiveClaimId', how='left')
)

# Filter for CPT code 99291
cpt_99291_df = merge_df[merge_df['HCPCS'] == '99291']

# Step 3: Find the most common diagnosis code
diagnosis_counts = (
    cpt_99291_df
    .groupby('CodeValue')
    .size()
    .reset_index(name='Count')
    .sort_values(by='Count', ascending=False)
)

print("Most common diagnosis codes associated with CPT 99291:")
print(diagnosis_counts.head(10))

print(f"\nThe most common diagnosis code is: {diagnosis_counts.iloc[0]['CodeValue']}")
print(f"It appears {diagnosis_counts.iloc[0]['Count']} times")
```

```python
#Question 8: Charges by Payer
```

```python
# Calculate total charges per claim and payer metrics
payer_summary = (
    merge_df
    .groupby(['ProspectiveClaimId', 'PrimaryPayerName'])
    .agg(TotalClaimCharges=('Charges', 'sum'))
    .reset_index()
    .groupby('PrimaryPayerName')
    .agg(
        TotalCharges=('TotalClaimCharges', 'sum'),
        AvgChargesPerClaim=('TotalClaimCharges', 'mean'),
        NumberOfClaims=('ProspectiveClaimId', 'count')
    )
    .reset_index()
    .sort_values(by='TotalCharges', ascending=False)
    .head(10)
)

print("Top 10 Payers by Total Charges:")
print("=" * 80)
print(payer_summary)
```

```python
# Analysis: Place of Service vs Average Charges
place_of_service_analysis = (
    merge_df
    .groupby(['ProspectiveClaimId', 'PlaceOfService_y'])
    .agg(TotalClaimCharges=('Charges', 'sum'))
    .reset_index()
    .groupby('PlaceOfService_y')
    .agg(
        TotalCharges=('TotalClaimCharges', 'sum'),
        AvgChargesPerClaim=('TotalClaimCharges', 'mean'),
        MedianChargesPerClaim=('TotalClaimCharges', 'median'),
        NumberOfClaims=('ProspectiveClaimId', 'count'),
        MinCharges=('TotalClaimCharges', 'min'),
        MaxCharges=('TotalClaimCharges', 'max')
    )
    .reset_index()
    .sort_values(by='AvgChargesPerClaim', ascending=False)
)

print("Place of Service Analysis - Sorted by Average Charges:")
print("=" * 100)
print(place_of_service_analysis)

# Additional insights
print("\n\nKey Insights:")
print(f"Highest average charges from place of service: {place_of_service_analysis.iloc[0]['PlaceOfService_y']} "
      f"(${place_of_service_analysis.iloc[0]['AvgChargesPerClaim']:,.2f})")
print(f"Lowest average charges from place of service: {place_of_service_analysis.iloc[-1]['PlaceOfService_y']} "
      f"(${place_of_service_analysis.iloc[-1]['AvgChargesPerClaim']:,.2f})")
print(f"Most common place of service: {place_of_service_analysis.loc[place_of_service_analysis['NumberOfClaims'].idxmax(), 'PlaceOfService_y']} "
      f"({place_of_service_analysis['NumberOfClaims'].max()} claims)")
```

```python
plot_df = (
    place_of_service_analysis[
        place_of_service_analysis['AvgChargesPerClaim'] > 0
    ]
    .sort_values(by='AvgChargesPerClaim', ascending=False)
)

# Convert x-axis values to strings to treat them as categories (if not converted you will see extra variables)
x_labels = plot_df['PlaceOfService_y'].astype(str)

plt.figure(figsize=(10, 6))
bars = plt.bar(x_labels, plot_df['AvgChargesPerClaim'], color='steelblue')

# Add labels on top of bars
for bar, value in zip(bars, plot_df['AvgChargesPerClaim']):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height(),
             f"${value:,.0f}", ha='center', va='bottom', fontsize=9)

plt.title("Average Claim Charges by Place of Service", fontsize=14, weight='bold')
plt.xlabel("Place of Service", fontsize=12)
plt.ylabel("Average Charges per Claim ($)", fontsize=12)
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```
